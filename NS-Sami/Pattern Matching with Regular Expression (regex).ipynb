```python
# without Regular Expression

def isPhoneNumber(text):
    if len(text)!= 12:
        return False
    for i in range(0, 3):
        if not text[i].isdecimal():
            return False
    if text[3] != '-':
        return False
    for i in range(4, 7):
        if not text[i].isdecimal():
            return False
    if text[7]!= '-':
       return False
    for i in range(8,12):
        if not text[i].isdecimal():
            return False
    return True

# print(isPhoneNumber('444-222-2525'))

message = "what is righ is right, you can't alter it no matter waht you do. the number of the affected person is 444-222-3333. i want you to call and appologize, also give him my number which is 222-333-6666. Pattern Matching with Regular Expression (regex).ipynb"

for i in range(len(message)):
    chunk = message[i:i+12]
    if isPhoneNumber(chunk):
        print("Phone Number Found: "+ chunk)
print ('Done')



```

```python
# With Regular Expression using search()

import re

message = "The number of the affected person is 444-222-3333.Give him my number which is 222-333-6666"

phoneNumRegex = re.compile(r'\d\d\d-\d\d\d-\d\d\d\d')
matchingObject = phoneNumRegex.search(message) # search only find the first match and exit
print('Phone Number Found: ' + matchingObject.group())

```

```python
# With Regular Expression using findall()
import re

message = "The number of the affected person is 444-222-3333.Give him my number which is 222-333-6666"

phoneNumRegex = re.compile(r'\d\d\d-\d\d\d-\d\d\d\d')
matchingObject = phoneNumRegex.findall(message) # findall()
print('Phone Number Found: ', matchingObject)
```

```python
# Grouping with parrenthesis

import re
message = "The number of the affected person is 444-222-3333.Give him my number which is 222-333-6666"
phoneNumRegex = re.compile(r'(\d\d\d)-(\d\d\d-\d\d\d\d)')
mo = phoneNumRegex.search(message)

print('Group 1: '+ mo.group(1))
print('Group 2: '+ mo.group(2))
print('Group 0: '+ mo.group(0))
print('Group : '+ mo.group())
print('Groups : ', mo.groups())
```
